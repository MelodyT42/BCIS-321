```python
# Visualization 1
from matplotlib import pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/MelodyT42/BCIS-321/refs/heads/main/shot_log-Melody_T.csv')
sns.histplot(data=df, x='TOUCH_TIME', bins=25, kde=False, color='r')
sns.despine()
plt.show()
```

```python
# Visualization 2
from matplotlib import pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/MelodyT42/BCIS-321/refs/heads/main/shot_log-Melody_T.csv')
sns.boxplot(data=df, x='SHOT_CLOCK', color='green')
plt.title('Box Plot of Shot Clock')
plt.xlabel('Shot Clock')
sns.despine()
plt.show()
```

```python
# Visualization 3
from matplotlib import pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/MelodyT42/BCIS-321/refs/heads/main/shot_log-Melody_T.csv')
ax = sns.boxplot(
    x=df.SHOT_RESULT,            
    y=df.SHOT_DIST,         
    hue=df.SHOT_RESULT,       
    palette={"made": "blue", "missed": "red"}, 
    fliersize=4, 
    saturation=0.50, 
    width=0.50, 
    linewidth=0.5
)
plt.title('Box Plot of Shot Result', fontsize=18)
plt.xlabel('Shot Result')
plt.ylabel('Count')
plt.show()
```
